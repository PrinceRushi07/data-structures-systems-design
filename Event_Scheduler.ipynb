```python

import heapq

def can_attend_all(events):
    events = sorted(events)

    for i in range(1, len(events)):
        previous_end = events[i - 1][1]
        current_start = events[i][0]

        if current_start < previous_end:
            return False

    return True
```

```python
def min_rooms_required(events):
    if not events:
        return 0

    events = sorted(events)
    rooms = []
    max_rooms = 0

    for start, end in events:
        # Remove all rooms whose meetings have ended.
        while rooms and rooms[0] <= start:
            heapq.heappop(rooms)

        heapq.heappush(rooms, end)

        max_rooms = max(max_rooms, len(rooms))

    return max_rooms
```

```python

events1 = [(9, 10), (10, 11), (11, 12)]

print("Example 1")
print("Can attend all:", can_attend_all(events1))
print("Minimum rooms:", min_rooms_required(events1))
```
